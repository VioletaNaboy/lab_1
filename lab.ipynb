### Завдання 1. Побудова лінійного графіка

Побудуйте графік функції $y = 2x + 1$ для $x$ від $-5$ до $5$ із кроком $1$. Додайте назву графіка, підписи осей $x$ та $y$, а також сітку.

```python
import matplotlib.pyplot as plt

x = [i for i in range (-5,6)]

y = [2*i+1 for i in x]

plt.figure()
plt.plot(x, y, marker="o", label="y=2x+1", )

plt.axhline(linewidth=0.8)
plt.axvline(linewidth=0.8)

plt.grid(True,linestyle="dotted", alpha=0.5)

plt.title("Графік функції $y = 2x + 1$")
plt.xlabel("x")
plt.ylabel("y")
plt.legend()

plt.show()
```

### Завдання 2. Графік квадратичної функції

Побудуйте графік $y = x^2 - 4x + 3$ на відрізку $[-2; 6]$ із кроком $0{,}1$. Задайте червоний колір і пунктирний стиль лінії. Додайте назву, підписи осей та сітку.

```python
import matplotlib.pyplot as plt
import numpy as np

x = np.arange(-2, 6.1, 0.1)

y = x**2 - 4*x + 3

plt.figure()
plt.plot(x, y, color="red", linestyle="dashed", label="$y = x^2 - 4x + 3$")

plt.title("Парабола $y = x^2 - 4x + 3$")
plt.xlabel("x")
plt.ylabel("y")
plt.grid(True, linestyle="dotted", alpha=0.5)

plt.axhline(0, linewidth=0.8)
plt.axvline(0, linewidth=0.8)

plt.legend()
plt.show()
```

### Завдання 3. Графік тригонометричної функції

Побудуйте графік $y = \sin(x)$ на відрізку $[-2\pi; 2\pi]$, використавши не менше ніж 200 точок. Встановіть синій колір лінії, підпишіть осі та додайте горизонтальну лінію $y = 0$.

```python
import matplotlib.pyplot as plt
import numpy as np

x = np.linspace(-2 * np.pi, 2 * np.pi, 200)
y = np.sin(x)

plt.figure(figsize=(8, 5))
plt.plot(x, y, color="blue")

plt.axhline(0, color="black", linewidth=0.8)

plt.xlabel('x')
plt.ylabel('y')

plt.show()
```

### Завдання 4. Дві функції на одному рисунку

На одній координатній площині побудуйте $y = \sin(x)$ і $y = \cos(x)$ для $x \in [0; 2\pi]$. Використайте різні кольори та стилі ліній. Додайте легенду, назву графіка й сітку.

```python
import matplotlib.pyplot as plt
import numpy as np

x = np.linspace(0, 2 * np.pi, 200)

y_sin = np.sin(x)
y_cos = np.cos(x)

plt.figure(figsize=(9, 6))

plt.plot(x, y_sin, color="blue", linestyle="solid", linewidth=2, label="y = sin(x)")

plt.plot(x, y_cos, color="red", linestyle="dashed", linewidth=2, label="y = cos(x)")

plt.title("Графіки функцій y = sin(x) та y = cos(x)", fontsize=14)
plt.xlabel("x", fontsize=12)
plt.ylabel("y", fontsize=12)
plt.grid(True, linestyle=':', alpha=0.6)
plt.legend(fontsize=12)

plt.axhline(0, color="black", linewidth=0.8, alpha=0.7)
plt.axvline(0, color="black", linewidth=0.8, alpha=0.7)

plt.show()
```

### Завдання 5. Графік із маркерами та підписом точки

### Завдання 7. Вертикальна стовпчикова діаграма

### Завдання 15. Зафарбування ділянок під графіком

Побудуйте $y = \sin(x)$ на відрізку $[0; 2\pi]$ (не менше ніж 200 точок). Зафарбуйте площу між кривою та віссю $Ox$: зеленим там, де $y \geq 0$, і червоним там, де $y < 0$. Додайте легенду та сітку.

```python
import matplotlib.pyplot as plt
import numpy as np

x = np.linspace(0, 2 * np.pi, 200)
y = np.sin(x)

plt.figure()
plt.plot(x, y, color="black", label="y = sin(x)")
plt.fill_between(x, y, 0, where=(y >= 0), color="green", alpha=0.5, label="y >= 0")
plt.fill_between(x, y, 0, where=(y < 0), color="red", alpha=0.5, label="y < 0")

plt.title("График функции y = sin(x)")
plt.xlabel("x")
plt.ylabel("y")
plt.axhline(0, color="gray", linewidth=2)
plt.axvline(0, color="gray", linewidth=2)
plt.grid(True)
plt.legend()
plt.show()
```

### Завдання 16. Два графіки поруч (subplots)

Створіть рисунок із двома підграфіками в одному рядку. Ліворуч побудуйте $y = \sin(x)$, праворуч — $y = \cos(x)$ для $x \in [0; 2\pi]$. Кожному графіку задайте власну назву та підписи осей, а всьому рисунку — спільний заголовок.

### Завдання 17. Сітка з чотирьох підграфіків

У сітці $2 \times 2$ побудуйте функції $y = x$, $y = x^2$, $y = x^3$ та $y = |x|$ на відрізку $[-2;2]$. Задайте кожному підграфіку назву, окремий колір лінії та сітку. Забезпечте відсутність накладання підписів.

### Завдання 19. Теплова карта матриці (heatmap)

Побудуйте теплову карту матриці кореляцій чотирьох ознак A, B, C, D:

$$
\begin{pmatrix}
1 & 0{,}7 & -0{,}4 & 0{,}2 \\
0{,}7 & 1 & -0{,}1 & 0{,}5 \\
-0{,}4 & -0{,}1 & 1 & -0{,}6 \\
0{,}2 & 0{,}5 & -0{,}6 & 1
\end{pmatrix}
$$

Підпишіть рядки та стовпці, задайте діапазон кольорів від $-1$ до $1$, додайте колірну шкалу й числові значення в усі клітинки.

```python
import matplotlib.pyplot as plt
import numpy as np

matrix = np.array([
    [1, 0.7, -0.4, 0.2],
    [0.7, 1, -0.1, 0.5],
    [-0.4, -0.1, 1, -0.6],
    [0.2, 0.5, -0.6, 1]
])
labels = ["A", "B", "C", "D"]

plt.figure()
im = plt.imshow(matrix, cmap="coolwarm", vmin=-1, vmax=1)
plt.colorbar(im)

plt.xticks(np.arange(len(labels)), labels)
plt.yticks(np.arange(len(labels)), labels)

for i in range(len(labels)):
    for j in range(len(labels)):
        plt.text(j, i, str(matrix[i, j]), ha="center", va="center", color="black")

plt.title("Теплова карта матриці кореляцій")
plt.show()
```

### Завдання 20. Контурна карта функції двох змінних

Для $x, y \in [-\pi; \pi]$ побудуйте заповнену контурну карту функції $z = \sin(x)\cos(y)$. Сформуйте прямокутну сітку щонайменше зі 100 значень за кожною змінною; використайте не менше ніж 15 рівнів, додайте колірну шкалу, заголовок і підписи осей.

```python
import matplotlib.pyplot as plt
import numpy as np

x = np.linspace(-np.pi, np.pi, 200)
y = np.linspace(-np.pi, np.pi, 200)
X, Y = np.meshgrid(x, y)
Z = np.sin(X) * np.cos(Y)

plt.figure()
contour = plt.contourf(X, Y, Z, levels=15, cmap="viridis")
plt.colorbar(contour)
plt.axhline(0, color="black", linewidth=0.5)
plt.axvline(0, color="black", linewidth=0.5)
plt.title("Контурна карта функції z = sin(x)cos(y)")
plt.xlabel("x")
plt.ylabel("y")
plt.show()
```
